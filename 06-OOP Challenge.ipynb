# Object Oriented Programming Challenge

For this challenge, create a bank account class that has two attributes:

* owner
* balance

and two methods:

* deposit
* withdraw

As an added requirement, withdrawals may not exceed the available balance.

Instantiate your class, make several deposits and withdrawals, and test to make sure the account can't be overdrawn.

```python
class Account:
    
    def __init__(self,owner,balance):
        self.owner = owner
        self.balance = balance
    def __str__(self):
        return ("Account owner: {} \nAccout balance: {}".format(self.owner,self.balance))

    def deposite(self,cst_deposit):
        self.balance = self.balance + cst_deposit
        print ("Deposite accepted")

    def withdraw(self,with_draw):
        if self.balance < with_draw:
            print ("Funds unavailable")
        else:
            self.balance = self.balance - with_draw
            print ("Withdraw accepted")
```

```python
# 1. Instantiate the class
acct1 = Account('Jose',100)
```

```python
# 2. Print the object
print(acct1)
```

```python
# 3. Show the account owner attribute
acct1.owner
```

```python
# 4. Show the account balance attribute
acct1.balance
```

```python
# 5. Make a series of deposits and withdrawals
acct1.deposite(50)
```

```python
print(acct1.balance)
```

```python
acct1.withdraw(75)
```

```python
# 6. Make a withdrawal that exceeds the available balance
acct1.withdraw(500)
```
